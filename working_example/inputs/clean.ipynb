```python
df = pd.read_csv('annual_summary_2025.csv')
```

```python
df1= pd.read_csv('monthly_summary_2025.csv')
```

```python
fuel_type_mapping = {
    'AG-ASTK99': 'AGGREGATED',
    'AG-BSTK99': 'AGGREGATED',
    'AG-DSTK99': 'AGGREGATED',
    'AG-ESTK06': 'AGGREGATED',
    'AG-ESTK99': 'AGGREGATED',
    'AG-FSTK99': 'AGGREGATED',
    'AG-JSTK99': 'AGGREGATED',
    'AG-LEDG02': 'AGGREGATED',
    'AG-LSTK05': 'AGGREGATED',
    'AG-LSTK99': 'AGGREGATED',
    'CMSTW-2': 'WIND',
    'IBD-ARTH1': 'INTERCONNECTOR',
    'IVD-ARTH1': 'INTERCONNECTOR',
    'KILSD-1': 'BATTERY',
    'KILSD-2': 'BATTERY',
    'KILSD-3': 'BATTERY',
    'KILSD-4': 'BATTERY',
    'PGPET-1': 'DIESEL'
}
```

```python
def update_fuel_type(df):
    for bmu, fuel in fuel_type_mapping.items():
        df.loc[
            (df['nationalGridBmUnit'] == bmu) & (df['Fuel_type'] == 'NAN'),
            'Fuel_type'
        ] = fuel
    return df
```

```python
# Update both DataFrames
df = update_fuel_type(df)
df1 = update_fuel_type(df1)
```
